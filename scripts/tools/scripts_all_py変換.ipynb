```python
import subprocess
import sys
from pathlib import Path


# ==================================================
# パス設定
# ==================================================

try:
    BASE_DIR = Path(__file__).resolve().parent
except NameError:
    BASE_DIR = Path.cwd()

# scripts フォルダ
SCRIPTS_DIR = BASE_DIR.parent

print(f"[INFO] 対象フォルダ: {SCRIPTS_DIR}")


# ==================================================
# 設定
# ==================================================

# このNotebook自身
SELF_NOTEBOOK = "py変換.ipynb"


# ==================================================
# 変換
# ==================================================

for notebook_path in sorted(
    SCRIPTS_DIR.rglob("*.ipynb")
):

    # .ipynb_checkpoints は除外
    if ".ipynb_checkpoints" in notebook_path.parts:
        continue

    # 自分自身は除外
    if notebook_path.name == SELF_NOTEBOOK:
        continue

    print(f"[INFO] 変換: {notebook_path.relative_to(SCRIPTS_DIR)}")

    subprocess.run(
        [
            sys.executable,
            "-m",
            "jupyter",
            "nbconvert",
            "--to",
            "script",
            str(notebook_path),
            "--output-dir",
            str(notebook_path.parent),
        ],
        check=True,
    )

print("[INFO] 完了")
```
